```python
import numpy as np
import pandas as pd
```

```python
pd.set_option('display.max_rows' , None);
pd.set_option('display.max_columns' , None);
```

```python
df = pd.read_csv('flats.csv')
df.sample(5)
```

```python
print(df.duplicated().sum());
```

```python
print(df.isnull().sum());
```

```python
df.drop(columns = ['link' , 'property_id'] , inplace = True);
```

```python
df.head()
```

```python
df.rename(columns = {'area':'price_per_sqft'} , inplace = True);
df.head();
```

```python
df.society.value_counts().shape
```

```python
import re
df['society'] = df['society'].apply(lambda name : re.sub(r'\d+(\.\d+)?\s?★','',str(name)).strip()).str.lower();
```

```python
df['society'].value_counts()
```

```python
df = df[df['price'] != 'Price on Request']
```

```python
def treat_price(x):
    if type(x) == float:
        return x;
    else:
        if x[1] == 'Lac':
            return round(float(x[0])/100 , 2);
        else:
            return round(float(x[0]),2);
```

```python
df['price'] = df['price'].str.split(' ').apply(treat_price);
```

```python
df.head()
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹' , '').str.replace(',' , '').str.strip().astype('float');
```

```python
df.head(5)
```

```python
df['areaWithType'].value_counts()
```

```python
df['bedRoom'].value_counts()
```

```python
df[df['bedRoom'].isna()]
```

```python
df = df[~df['bedRoom'].isnull()]
```

```python
df['bedRoom'] = df['bedRoom'].str.split(' ').str.get(0).astype('int');
```

```python
df.head()
```

```python
df['bathroom'] = df['bathroom'].str.split(' ').str.get(0).astype('int');
```

```python
df['balcony'] = df['balcony'].str.split(' ').str.get(0).str.replace('No','0');
```

```python
df.head()
```

```python
df['additionalRoom'].value_counts()
```

```python
df['additionalRoom'].isna().sum()
```

```python
df['additionalRoom'].fillna('not avilable' , inplace = True)
```

```python
df['additionalRoom'].isna().sum()
```

```python
df['additionalRoom'] = df['additionalRoom'].str.lower()
```

```python
df.head()
```

```python
df['floorNum'] = df['floorNum'].str.split(' ').str.get(0)

replacements = {
    'Ground': '0',
    'Basement': '-1', 
    'Lower': '0'
}

df['floorNum'] = df['floorNum'].replace(replacements)
df['floorNum'] = df['floorNum'].str.extract(r'(-?\d+)', expand=False)
df['floorNum'] = pd.to_numeric(df['floorNum'], errors='coerce').fillna(0).astype(int)
```

```python
df.head()
```

```python
df['facing'].fillna('NA' , inplace = True);
```

```python
df['facing'].isna().sum()
```

```python
df.insert(loc = 4 , column = 'area' , value = round((df['price']*10000000)/df['price_per_sqft']));
```

```python
df.insert(loc = 1 , column = 'prooerty_type' , value = 'flat');
```

```python
df.to_csv('flats_cleaned.csv' , index = False);
```
